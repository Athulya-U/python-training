```python
n=int(input("enter number of elements"))
l=[]
for i in range(n):
  m=int(input("enter the element"))
  l.append(m)
  print(l)
  print("the largest element is",max(l))
```

```python
n=int(input("enter number of elements"))
l=[]
for i in range(n):
  m=int(input("enter the element"))
  l.append(m)
  print(l)
  print("the smallest element is",min(l))
```

```python
n=int(input("Enter the number of elements"))
sum=0
for i in range(n):
  l=int(input("Enter the element"))
  sum=sum+l
  print("The sum of elements is",sum)
```
